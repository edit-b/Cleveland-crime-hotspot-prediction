## 1. Cleaning the crime data

```python
import pandas as pd    
import glob
import os
from itertools import product
```

### Paths

```python

script_dir=os.path.dirname(os.path.abspath('__file__'))
project_dir=os.path.dirname(script_dir)
data_dir=os.path.join(project_dir, 'Data')

input_path=os.path.join(data_dir, 'Cleveland', '*', '*.csv')
output_path=os.path.join(data_dir, 'crime_monthly_lsoa.csv')
```

### Loading the crime data & dropping 'Context' column

```python

files=sorted(glob.glob(input_path))

df=pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
print(f"Loaded: {len(df):,} records from {len(files):,} monthly files")
print(f"Date range: {df['Month'].min()} to {df['Month'].max()}")

# Dropping  unused 'Context' coumn
df=df.drop(columns=['Context'])
print(f"Dropped 'Context' column")
```

### Initial data inspection

```python

print(f"Total records: {len(df):,}")
print(f"Unique LSOA codes: {df['LSOA code'].nunique()}")
print(f"Unique LSOA names: {df['LSOA name'].nunique()}")
print(f"Missing LSOA codes: {df['LSOA code'].isna().sum()}")
print(f"Missing Crime IDs: {df['Crime ID'].isna().sum():,} --> ASB records have no ID")

print(f"\nCrime type breakdown:")
for ct, count in df['Crime type'].value_counts().items():
    print(f" {ct}: {count:,} ({count/len(df)*100:.1f}%)")
```

### Extracting Local Authority and filtering to Teesside

```python

df['Local_Authority']=df['LSOA name'].str.rsplit(' ', n=1).str[0]

core_las=['Hartlepool', 'Middlesbrough', 'Redcar and Cleveland', 'Stockton-on-Tees']
outside=df[~df['Local_Authority'].isin(core_las)]
print(f"Records outside core area {len(outside):,}")
print(f"From: {outside['Local_Authority'].value_counts().to_dict()}")

df=df[df['Local_Authority'].isin(core_las)].copy()
print(f"After filtering: {len(df):,} records, {df['LSOA code'].nunique()} LSOAs")
```

### Building LSOA lookup



```python
#Some LSOA codes have 2 names, the most common name per code will be used for consistency
lsoa_lookup=(df.groupby('LSOA code')['LSOA name'].agg(lambda x: x.value_counts().index[0]).reset_index())
lsoa_lookup['Local_Authority']=lsoa_lookup['LSOA name'].str.rsplit(' ', n=1).str[0]

name_variations=df.groupby('LSOA code')['LSOA name'].nunique()
multi_name=name_variations[name_variations >1]
if len(multi_name)>0:
    print(f" {len(multi_name)} LSOA codes had multiple names --> using most common name")
else: 
    print(f"\nNo LSOA name variations found")

```

```python
#Aggregating to monthly counts per LSOA 
monthly=df.groupby(['Month', 'LSOA code']).size().reset_index(name='crime_count')
print(f"Aggregated to monthly conts:")
print(f"Rows: {len(monthly):,}")
```

```python
#Filling missing LSOA-month combinations with zero 
all_months=sorted(monthly['Month'].unique())
all_lsoas=sorted(monthly['LSOA code'].unique())
expected=len(all_months)*len(all_lsoas)
missing=expected-len(monthly)

full_index=pd.DataFrame(list(product(all_months, all_lsoas)), columns=['Month', 'LSOA code'])
monthly=full_index.merge(monthly, on=['Month', 'LSOA code'], how='left')
monthly['crime_count']=monthly['crime_count'].fillna(0).astype(int)

#Attaching LSOA name and LOcal Authority
monthly=monthly.merge(lsoa_lookup, on='LSOA code', how='left')

print(f"Expected combinations:{expected:,}, {len(all_months)} months x {len(all_lsoas)} LSOAs)")
print(f"Missing combination filled with 0: {missing:,}")
print(f"Final row count: {len(monthly):,}")

```

### Summary

```python

dupes=monthly.duplicated(subset=['Month', 'LSOA code']).sum()
print(f"Duplicate LSOA-month combos: {dupes}")
print(f"Zero-count LSOA-months: {(monthly['crime_count']==0).sum()} ({(monthly['crime_count']==0).mean()*100:.1f}%)")
print(f"Crime count-Min: {monthly['crime_count'].min()}, Max: {monthly['crime_count'].max()},"
      f"MEan: {monthly['crime_count'].mean():.1f}, Median: {monthly['crime_count'].median():.1f}")


```

### Saving

```python

monthly=monthly.sort_values(['LSOA code', 'Month']).reset_index(drop=True)
monthly.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
print(f"Rows: {len(monthly):,}")
print(f"Columns: {list(monthly.columns)}")
```
